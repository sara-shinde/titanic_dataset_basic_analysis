Now, we should see if survival rates vary by age, class, and sex.

```python
dataset.groupby(["Pclass"])["Survived"].value_counts(normalize=True)
```

These are the survival rates by Pclass (a more concise way to present information in the above cell)
.mean() instead of value_counts(normalize=True) works here because for this dataset, Survived column contains 0s and 1s.

```python
df = dataset.groupby(["Pclass"])["Survived"].mean()
```

```python
df
```

```python
df.plot(kind="bar")
plt.title("Survival Rate by Passenger Class")
plt.ylabel("Survival rate")
plt.xlabel("Passenger class")
plt.show()
```

```python
df2 = dataset.groupby(["Sex"])["Survived"].mean()
```

```python
df2
```

```python
df2.plot(kind="bar")
plt.title("Survival Rate by Sex")
plt.ylabel("Survival rate")
plt.xlabel("Sex")
plt.show()
```

Let us create a new column called age_groups with the following age brackets: [0, 16), [16, +inf)

```python
age_bins = [0, 16, 150]
age_labels = ["child", "adult"]
```

```python
dataset["AgeGroup"] = pd.cut(dataset["Age"], bins=age_bins, labels=age_labels, right=False)
```

```python
dataset.head()
```

```python
dataset.AgeGroup.value_counts()
```

```python
df3 = dataset.groupby(["AgeGroup"])["Survived"].mean()
```

```python
df3
```

```python
df3.plot(kind="bar")
plt.title("Survival rate by Age Group")
plt.ylabel("Survival rate")
plt.xlabel("Age group")
plt.show()
```

Now, let us see how the intersection of age, class, and sex impacts survival.

```python
df4 = dataset.groupby(["Sex", "Pclass"])["Survived"].mean()
```
